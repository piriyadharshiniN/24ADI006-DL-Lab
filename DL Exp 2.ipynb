```python
print("Piriyadharshini N\n24BAD086\n")

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import tensorflow as tf

from tensorflow.keras.datasets import fashion_mnist
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Flatten
from tensorflow.keras.callbacks import EarlyStopping
from tensorflow.keras.optimizers import SGD, RMSprop, Adam

print("TensorFlow Version :", tf.__version__)

# Load dataset
(x_train, y_train), (x_test, y_test) = fashion_mnist.load_data()
print("\nDataset Loaded Successfully")

# Dataset info
print("\nTraining Images :", x_train.shape)
print("Training Labels :", y_train.shape)
print("\nTesting Images :", x_test.shape)
print("Testing Labels :", y_test.shape)
print("\nNumber of Classes :", len(np.unique(y_train)))

# Normalize dataset
x_train = x_train / 255.0
x_test = x_test / 255.0
print("\nDataset Normalized Successfully")

# Display sample images
class_names = [
    'T-shirt','Trouser','Pullover','Dress','Coat',
    'Sandal','Shirt','Sneaker','Bag','Ankle Boot'
]

plt.figure(figsize=(8,8))
for i in range(16):
    plt.subplot(4,4,i+1)
    plt.imshow(x_train[i], cmap='gray')
    plt.title(class_names[y_train[i]], fontsize=8)
    plt.axis("off")
plt.suptitle("Fashion-MNIST Sample Images")
plt.show()

# Create ANN model
def create_model(optimizer):
    model = Sequential()
    model.add(Flatten(input_shape=(28,28)))
    model.add(Dense(256, activation='relu'))
    model.add(Dense(128, activation='relu'))
    model.add(Dense(64, activation='relu'))
    model.add(Dense(10, activation='softmax'))
    model.compile(optimizer=optimizer,
                  loss='sparse_categorical_crossentropy',
                  metrics=['accuracy'])
    return model

# Early stopping
early_stop = EarlyStopping(monitor='val_loss',
                           patience=3,
                           restore_best_weights=True)

# Define optimizers
optimizers = {
    "SGD" : SGD(learning_rate=0.01),
    "Momentum" : SGD(learning_rate=0.01, momentum=0.9),
    "RMSProp" : RMSprop(learning_rate=0.001),
    "Adam" : Adam(learning_rate=0.001)
}

# Train models
models = {}
histories = {}
for name, optimizer in optimizers.items():
    print("\nTraining using", name)
    model = create_model(optimizer)
    model.summary()
    history = model.fit(x_train, y_train,
                        epochs=20,
                        batch_size=128,
                        validation_split=0.2,
                        callbacks=[early_stop],
                        verbose=1)
    models[name] = model
    histories[name] = history

print("\nAll Optimizers Trained Successfully.")

# Save histories
history_sgd = histories["SGD"]
history_momentum = histories["Momentum"]
history_rmsprop = histories["RMSProp"]
history_adam = histories["Adam"]

# Display final training/validation accuracy
print("\nFinal Training Accuracy")
print("SGD       :", history_sgd.history['accuracy'][-1])
print("Momentum  :", history_momentum.history['accuracy'][-1])
print("RMSProp   :", history_rmsprop.history['accuracy'][-1])
print("Adam      :", history_adam.history['accuracy'][-1])

print("\nFinal Validation Accuracy")
print("SGD       :", history_sgd.history['val_accuracy'][-1])
print("Momentum  :", history_momentum.history['val_accuracy'][-1])
print("RMSProp   :", history_rmsprop.history['val_accuracy'][-1])
print("Adam      :", history_adam.history['val_accuracy'][-1])

# Evaluate models
results = {}
for name, model in models.items():
    loss, accuracy = model.evaluate(x_test, y_test, verbose=0)
    results[name] = [loss, accuracy]
    print("\nOptimizer :", name)
    print("Test Loss :", loss)
    print("Test Accuracy :", accuracy)

# Training accuracy graph
plt.figure(figsize=(10,6))
for name in histories:
    plt.plot(histories[name].history['accuracy'], label=name)
plt.title("Training Accuracy vs Epoch")
plt.xlabel("Epoch")
plt.ylabel("Accuracy")
plt.legend()
plt.grid()
plt.show()

# Validation accuracy graph
plt.figure(figsize=(10,6))
for name in histories:
    plt.plot(histories[name].history['val_accuracy'], label=name)
plt.title("Validation Accuracy vs Epoch")
plt.xlabel("Epoch")
plt.ylabel("Validation Accuracy")
plt.legend()
plt.grid()
plt.show()

# Training loss graph
plt.figure(figsize=(10,6))
for name in histories:
    plt.plot(histories[name].history['loss'], label=name)
plt.title("Training Loss vs Epoch")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.legend()
plt.grid()
plt.show()

# Validation loss graph
plt.figure(figsize=(10,6))
for name in histories:
    plt.plot(histories[name].history['val_loss'], label=name)
plt.title("Validation Loss vs Epoch")
plt.xlabel("Epoch")
plt.ylabel("Validation Loss")
plt.legend()
plt.grid()
plt.show()

# Accuracy bar chart
optimizer_names = list(results.keys())
accuracy_values = [results[x][1] for x in optimizer_names]
plt.figure(figsize=(8,5))
plt.bar(optimizer_names, accuracy_values)
plt.title("Final Test Accuracy")
plt.ylabel("Accuracy")
plt.ylim(0.7,1.0)
plt.grid(axis='y')
plt.show()

# Loss bar chart
loss_values = [results[x][0] for x in optimizer_names]
plt.figure(figsize=(8,5))
plt.bar(optimizer_names, loss_values)
plt.title("Final Test Loss")
plt.ylabel("Loss")
plt.grid(axis='y')
plt.show()

# Predictions
predictions = {}
for name, model in models.items():
    prediction = model.predict(x_test, verbose=0)
    predictions[name] = np.argmax(prediction, axis=1)

# Confusion matrices
from sklearn.metrics import confusion_matrix, classification_report
for name in models:
    cm = confusion_matrix(y_test, predictions[name])
    plt.figure(figsize=(8,6))
    plt.imshow(cm)
    plt.title(name + " Confusion Matrix")
    plt.colorbar()
    plt.xlabel("Predicted Label")
    plt.ylabel("True Label")
    plt.show()

# Classification reports
for name in models:
    print("\n", name, "Classification Report")
    print(classification_report(y_test, predictions[name], target_names=class_names))

# Sample predictions (Adam)
plt.figure(figsize=(12,10))
for i in range(16):
    plt.subplot(4,4,i+1)
    plt.imshow(x_test[i], cmap="gray")
    plt.title("Actual : " + class_names[y_test[i]] +
              "\nPred : " + class_names[predictions["Adam"][i]], fontsize=8)
    plt.axis("off")
plt.suptitle("Sample Predictions using Adam Optimizer")
plt.show()

# Comparison table
comparison = pd.DataFrame({
    "Optimizer":optimizer_names,
    "Training Accuracy":[history_sgd.history['accuracy'][-1],
                         history_momentum.history['accuracy'][-1],
                         history_rmsprop.history['accuracy'][-1],
                         history_adam.history['accuracy'][-1]],
    "Validation Accuracy":[history_sgd.history['val_accuracy'][-1],
                           history_momentum.history['val_accuracy'][-1],
                           history_rmsprop.history['val_accuracy'][-1],
                           history_adam.history['val_accuracy'][-1]],
    "Training Loss":[history_sgd.history['loss'][-1],
                     history_momentum.history['loss'][-1],
                     history_rmsprop.history['loss'][-1],
                     history_adam.history['loss'][-1]],
    "Validation Loss":[history_sgd.history['val_loss'][-1],
                       history_momentum.history['val_loss'][-1],
                       history_rmsprop.history['val_loss'][-1],
                       history_adam.history['val_loss'][-1]],
    "Test Accuracy":[results["SGD"][1],
                     results["Momentum"][1],
                     results["RMSProp"][1],
                     results["Adam"][1]]
})

print("\nOptimizer Comparison Table\n")
print(comparison)

comparison.to_csv("Optimizer_Comparison.csv", index=False)
print("\nComparison table saved as Optimizer_Comparison.csv")

# Best optimizer
best_optimizer = comparison.loc[comparison["Test Accuracy"].idxmax(), "Optimizer"]
best_accuracy = comparison["Test Accuracy"].max()
print("\nBest Optimizer :", best_optimizer)
print("Best Test Accuracy :", best_accuracy)
```
